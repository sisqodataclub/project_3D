```python
def data_overview(data, title):
    overview_analysis = {f'{title}':[data.shape[1], data.shape[0], 
                                     data.isnull().any(axis=1).sum(), 
                                     data.isnull().any(axis=1).sum()/len(data)*100,
                                     data.duplicated().sum(),
                                    data.duplicated().sum()/len(data)*100, 
                                     sum((data.dtypes == 'object') & (data.nunique() > 2)),
                                     sum((data.dtypes == 'object') & (data.nunique() < 3)),
                                     data.select_dtypes(include=['int64', 'float64']).shape[1]
                                    ]}
    overview_analysis=pd.DataFrame(overview_analysis, index=['Columns','Rows','Missing_Values','Missing_Values %',
                                                             'Duplicates', 'Duplicates %','Categorical_variables','Boolean_variables','Numerical_variables']).round(2)
    return overview_analysis


def has_digits(s):
    """Return True if the string representation contains any digit."""
    return any(char.isdigit() for char in str(s))

def variables_overview(data):
    variable_details = {
        'unique': data.nunique(),
        'dtype': data.dtypes,
        'null': data.isna().sum(),
        'null %': data.isna().sum() / len(data) * 100,
    }
    variable_details = pd.DataFrame(variable_details)


    # Add a new column 'has_non_alphanumeric' to indicate if there are non-alphanumeric characters
    
    variable_details['has_non_alphanumeric'] = data.apply(lambda col: any(col.apply(lambda x: not str(x).replace(" ", "").isalnum())))
    #variable_details['has_non_alphanumeric'] = data.apply(lambda col: any(col.apply(lambda x: not str(x).isalnum())))

    # Add a new column 'has_digits' to indicate if there are rows containing digits in each column
    variable_details['has_digits'] = data.apply(lambda col: any(col.apply(has_digits)))

    
    return variable_details
```

```python
variables_overview(df_bronze)
```

```python
print("--- DEEP DIVE EDA ---")

# 1. Expose the 5 unique variations of 'sex'
print("\n1. Exact values in 'sex' column:")
print(df_bronze['sex'].value_counts())

# 2. Expose the exact negative and impossible ages
print("\n2. Out-of-bounds Ages (< 0 or > 120):")
bad_ages = df_bronze[(df_bronze['age'] < 0) | (df_bronze['age'] > 120)]
print(bad_ages['age'].value_counts().sort_index())

# 3. Expose Postcode casing inconsistencies
print("\n3. Postcode casing check (Sample of 10):")
print(df_bronze['postcode'].sample(10).values)
```

```python
print("--- APPLYING SILVER LAYER CLEANING ---")

# 1. Create a copy of the Bronze data to become our Silver data
df_silver = df_bronze.copy()
starting_count = len(df_silver)

# Rule 1: Clean the 'sex' column
# We create a dictionary to map all variations to the official 3 categories
sex_mapping = {
    'M': 'Male',
    'Male': 'Male',
    'F': 'Female',
    'Female': 'Female',
    'unknown': 'Unknown'
}
df_silver['sex'] = df_silver['sex'].map(sex_mapping)

# Rule 2: Clean the 'age' column
# We forcefully drop any rows where the age is impossible
df_silver = df_silver[(df_silver['age'] >= 0) & (df_silver['age'] <= 120)]

# Rule 3: Clean the 'postcode' column
# We force all characters to uppercase to fix things like 'mO0 1qt'
df_silver['postcode'] = df_silver['postcode'].str.upper()

# --- VERIFICATION ---
ending_count = len(df_silver)
dropped_rows = starting_count - ending_count

print(f"Starting Rows: {starting_count}")
print(f"Cleaned Rows:  {ending_count}")
print(f"Rows Dropped:  {dropped_rows} (due to invalid ages)")

print("\nVerify 'sex' column is fixed:")
print(df_silver['sex'].value_counts())

print("\nVerify 'age' boundaries:")
print(f"Min Age: {df_silver['age'].min()} | Max Age: {df_silver['age'].max()}")

print("\nVerify 'postcode' casing (Sample):")
print(df_silver['postcode'].sample(5).values)
```

```python
import pandas as pd
import pandas_gbq
from google.oauth2 import service_account

# 1. Point to your JSON key
key_path = "/home/kali5/new_folder/credentials/gcp-service-account.json"
credentials = service_account.Credentials.from_service_account_file(key_path)

PROJECT_ID = 'noble-resolver-389220'
DATASET_ID = 'ons_census'

print("🔍 Scanning BigQuery for all tables...\n")

# 2. Query BigQuery's internal schema to get a list of ALL tables in the dataset
schema_query = f"""
    SELECT table_name 
    FROM `{PROJECT_ID}.{DATASET_ID}.INFORMATION_SCHEMA.TABLES`
"""

try:
    tables_df = pandas_gbq.read_gbq(schema_query, project_id=PROJECT_ID, credentials=credentials)
    table_names = tables_df['table_name'].tolist()
    
    if not table_names:
        print("⚠️ No tables found in the dataset.")
    else:
        print(f"✅ Found {len(table_names)} tables: {', '.join(table_names)}\n")
        print("="*60)
        
        # 3. Loop through every table found and fetch 5 rows
        for table in table_names:
            print(f"📊 Table: {table.upper()}")
            
            data_query = f"SELECT * FROM `{PROJECT_ID}.{DATASET_ID}.{table}` LIMIT 5"
            df_data = pandas_gbq.read_gbq(data_query, project_id=PROJECT_ID, credentials=credentials)
            
            if df_data.empty:
                print("(Table is empty)\n")
            else:
                print(df_data.head())
                print("\n" + "="*60)

except Exception as e:
    print(f"❌ Error connecting to BigQuery: {e}")
```
