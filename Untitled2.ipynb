```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid")
```

```python
# The National Grid Carbon Intensity API endpoint
url = "https://api.carbonintensity.org.uk/generation"

print("Fetching live data from National Grid ESO...")
response = requests.get(url)

if response.status_code == 200:
    raw_data = response.json()
    print("✅ Success! Raw JSON data fetched.")
else:
    print(f"❌ Failed to fetch data. Status Code: {response.status_code}")
```

```python
# Extract the timestamp and the actual fuel array
timestamp = raw_data['data']['from']
fuel_mix = raw_data['data']['generationmix']

# Convert to Pandas DataFrame
df = pd.DataFrame(fuel_mix)

# Add the timestamp so we know exactly when this snapshot was taken
df['recorded_at'] = pd.to_datetime(timestamp)

# Sort by percentage so the biggest energy sources are at the top
df = df.sort_values(by='perc', ascending=False).reset_index(drop=True)

print(f"UK Grid Snapshot at: {timestamp}")
display(df)
```

```python
plt.figure(figsize=(12, 6))

# Create a bar chart
ax = sns.barplot(data=df, x='fuel', y='perc', palette='mako')

# Add labels and titles
plt.title(f'Live UK Energy Generation Mix (%)\nSnapshot: {timestamp}', fontsize=16, pad=15)
plt.xlabel('Fuel Source', fontsize=12)
plt.ylabel('Percentage of Grid (%)', fontsize=12)
plt.xticks(rotation=45, ha='right', fontsize=11)

# Add the exact percentages on top of each bar for readability
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', padding=3)

plt.tight_layout()
plt.show()
```
