```python
import pandas as pd
import requests
from io import BytesIO

def download_uk_business_population_data():
    """
    Downloads the latest Business Population Estimates data from data.gov.uk
    and loads it into a pandas DataFrame.
    """
    # URL for the 2025 edition (XLSX format) - you can replace with other editions
    file_url = "https://assets.publishing.service.gov.uk/media/68dbccc9c487360cc70c9f4e/BPE_2025_detailed_tables.xlsx"
    
    print(f"Downloading data from: {file_url}")
    
    # Download the file
    response = requests.get(file_url)
    response.raise_for_status()  # Raise an error for bad status codes
    
    # Load the Excel file into a pandas DataFrame
    # Note: The Excel file contains multiple sheets; you can specify sheet_name
    # Here we load the first sheet by default
    df = pd.read_excel(BytesIO(response.content), sheet_name=0, engine="openpyxl")
    
    print("Download and load successful!")
    return df

# Example usage
if __name__ == "__main__":
    df = download_uk_business_population_data()
    print(df.sample(15))
```
