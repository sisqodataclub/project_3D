```python
def data_overview(data, title):
    overview_analysis = {f'{title}':[data.shape[1], data.shape[0], 
                                     data.isnull().any(axis=1).sum(), 
                                     data.isnull().any(axis=1).sum()/len(data)*100,
                                     data.duplicated().sum(),
                                    data.duplicated().sum()/len(data)*100, 
                                     sum((data.dtypes == 'object') & (data.nunique() > 2)),
                                     sum((data.dtypes == 'object') & (data.nunique() < 3)),
                                     data.select_dtypes(include=['int64', 'float64']).shape[1]
                                    ]}
    overview_analysis=pd.DataFrame(overview_analysis, index=['Columns','Rows','Missing_Values','Missing_Values %',
                                                             'Duplicates', 'Duplicates %','Categorical_variables','Boolean_variables','Numerical_variables']).round(2)
    return overview_analysis


def has_digits(s):
    """Return True if the string representation contains any digit."""
    return any(char.isdigit() for char in str(s))

def variables_overview(data):
    variable_details = {
        'unique': data.nunique(),
        'dtype': data.dtypes,
        'null': data.isna().sum(),
        'null %': data.isna().sum() / len(data) * 100,
    }
    variable_details = pd.DataFrame(variable_details)


    # Add a new column 'has_non_alphanumeric' to indicate if there are non-alphanumeric characters
    
    variable_details['has_non_alphanumeric'] = data.apply(lambda col: any(col.apply(lambda x: not str(x).replace(" ", "").isalnum())))
    #variable_details['has_non_alphanumeric'] = data.apply(lambda col: any(col.apply(lambda x: not str(x).isalnum())))

    # Add a new column 'has_digits' to indicate if there are rows containing digits in each column
    variable_details['has_digits'] = data.apply(lambda col: any(col.apply(has_digits)))

    
    return variable_details
```

```python
import pandas as pd

def bronze_to_silver(df_bronze):
    df = df_bronze.copy()
    
    # 1. Drop redundant/constant columns
    df.drop(columns=['Time', 'uk-only', 'Geography'], inplace=True, errors='ignore')
    
    # 2. Rename columns for clarity
    df.rename(columns={
        'v4_0': 'index_value',
        'mmm-yy': 'period_str',
        'cpih1dim1aggid': 'item_code',
        'Aggregate': 'item_name'
    }, inplace=True)
    
    # 3. Convert period to proper date (first day of month)
    df['period'] = pd.to_datetime(df['period_str'], format='%b-%y')
    
    # 4. Clean item_name: strip whitespace, remove ellipsis if needed
    df['item_name'] = df['item_name'].str.strip().str.replace('...', '', regex=False)
    
    # 5. Flag zero values (potential data quality issue)
    df['is_zero'] = df['index_value'] == 0.0
    
    # 6. Sort for time-series efficiency
    df = df.sort_values(['item_code', 'period']).reset_index(drop=True)
    
    # 7. Set appropriate dtypes
    df['item_code'] = df['item_code'].astype('category')
    
    silver_df = df[['period', 'item_code', 'item_name', 'index_value', 'is_zero']]
    return silver_df
```

```python
import pandas as pd

# ============================================================================
# SILVER LAYER: Enriched Fact Table (State, No Metrics)
# ============================================================================
# Merge silver_df with dimension attributes and weights
silver_enriched = silver_df.merge(
    dim_item[['item_code', 'division', 'division_name', 'weight_pp1000']],
    on='item_code',
    how='left'
)

# Optional: Save silver_enriched to CSV or database
# silver_enriched.to_csv('silver_enriched.csv', index=False)

print("Silver layer created: silver_enriched")
print(f"Shape: {silver_enriched.shape}")
print(silver_enriched.head())

# ============================================================================
# GOLD LAYER: Business Metrics & Aggregations (Read from silver_enriched)
# ============================================================================
# Note: In production, you would read silver_enriched from your database/parquet.
# Here we use it directly.

# ----------------------------------------------------------------------------
# PART A: HEADLINE CPI (using actual code 'CP00' from the data)
# ----------------------------------------------------------------------------
headline = silver_enriched[
    (silver_enriched['item_code'] == 'CP00') & 
    (~silver_enriched['is_zero'])
].copy()
headline = headline.sort_values('period')

# Year‑over‑year percentage change (12‑month shift) – this is a Gold metric
headline['headline_yoy_pct'] = headline['index_value'].pct_change(12) * 100

gold_headline = headline[['period', 'index_value', 'headline_yoy_pct']].rename(
    columns={'index_value': 'headline_index'}
)

# ----------------------------------------------------------------------------
# PART B: DIVISION‑LEVEL WEIGHTED INDICES (Bottom‑up from leaf items)
# ----------------------------------------------------------------------------
# Keep only detailed leaf items that have positive weights
detailed = silver_enriched[
    (silver_enriched['weight_pp1000'] > 0) & 
    (~silver_enriched['is_zero']) & 
    (silver_enriched['item_code'] != 'CP00')
].copy()

# Weighted average index for each division per period
division_index = detailed.groupby(['period', 'division_name']).apply(
    lambda g: (g['index_value'] * g['weight_pp1000']).sum() / g['weight_pp1000'].sum()
).reset_index(name='division_index')

# Pivot to wide format (each division becomes a column)
gold_divisions_index = division_index.pivot(
    index='period', columns='division_name', values='division_index'
)

# Year‑over‑year percentage changes for all divisions
gold_divisions_yoy = gold_divisions_index.pct_change(12) * 100

# ----------------------------------------------------------------------------
# COMBINED GOLD TABLE (Tidy long format, ready for dashboards)
# ----------------------------------------------------------------------------
gold_divisions_yoy_flat = gold_divisions_yoy.reset_index().melt(
    id_vars='period', var_name='division_name', value_name='yoy_pct'
)

gold_combined = gold_headline.merge(
    gold_divisions_yoy_flat, on='period', how='outer'
)

# Remove the first 12 months where YoY cannot be calculated
gold_combined = gold_combined.dropna(subset=['headline_yoy_pct']).reset_index(drop=True)

# ============================================================================
# DISPLAY & EXPORT RESULTS
# ============================================================================
print("\n--- Gold Headline CPI (Last 5 Months) ---")
print(gold_headline.tail())

print("\n--- Gold Division YoY % (Last 5 Months) ---")
print(gold_divisions_yoy.tail())

print("\n--- Gold Combined Table (Recent Data) ---")
print(gold_combined.tail(10))

# Optional: Save Gold tables to CSV for BI tools
# gold_headline.to_csv('gold_headline.csv', index=False)
# gold_divisions_yoy.to_csv('gold_divisions_yoy.csv')
# gold_combined.to_csv('gold_combined.csv', index=False)
```
